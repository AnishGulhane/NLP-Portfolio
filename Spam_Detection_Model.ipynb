```python
# we are yet to remove the stopwords
```

```python
df['X'] = df['X'].str.lower() # python is case sensitive thats why we are converting every thing in lower case or upper case 
```

```python
def text_process(mess):
    """
    1. remove the punctuation 
    2. remove the stopwords
    3. return the list of clean textwords
    """
    nopunc = [char for char in mess if char not in string.punctuation]
    nopunc = "".join(nopunc)

    return[ word for word in nopunc.split() if word not in stopwords.words('english')]
```

```python
df['X'].apply(text_process)
```

```python
# now we will convert every unique word into a number
```

```python
from sklearn.feature_extraction.text import CountVectorizer
# assign a ID/token number to each word
```

```python
bow_transformer = CountVectorizer(analyzer=text_process).fit(df["X"])
```

Cell [25] (bow_transformer = ...): This does the heavy lifting for setting up the vocabulary.

It creates a CountVectorizer and tells it to use your custom text_process function as the analyzer. This means before it counts anything, it runs your punctuation-and-stopword-removal code on the text.

The .fit(df["X"]) part tells it to scan the entire 'X' column to learn the complete vocabulary (every single unique clean word used across all messages). "BOW" stands for Bag of Words.

```python
len(bow_transformer.vocabulary_)
```

```python
# tdm --> term document matrix
```

```python
print(len(bow_transformer.vocabulary_))
```

```python
tdm = bow_transformer.transform(df['X'])
```

Cell [29]: Converts the raw text column (df['X']) into a term-document matrix (tdm), a numerical grid where values represent word counts.

```python
tdm.shape
```

Cell [30]: Outputs the matrix dimensions, confirming 6,776 rows (individual text messages) and 9,422 columns (unique words).

```python
from sklearn.model_selection import train_test_split
train_x, test_x, train_y, test_y = train_test_split(tdm, df.Y)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score,recall_score,precision_score,f1_score
from sklearn.metrics import *
```

```python
train_y = train_y.astype(int)
test_y = test_y.astype(int)
```

Convert target variables to integers to ensure compatibility with machine learning models

```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression()
logreg.fit(train_x, train_y)#model build
```

```python
pred = logreg.predict(test_x)
print(confusion_matrix(test_y,pred))
print(classification_report(test_y, pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier()
dt.fit(train_x, train_y)
pred = dt.predict(test_x)
tab = confusion_matrix(test_y, pred)
print(tab)
print(
    accuracy_score(test_y, pred),
    " ",
    recall_score(test_y, pred),
    " ",
    precision_score(test_y, pred))
```

```python

from  sklearn.ensemble import RandomForestClassifier
dt = RandomForestClassifier(oob_score=True)
dt.fit(train_x,train_y)
pred = dt.predict(test_x)
tab = confusion_matrix(test_y, pred)
print(tab)
print(accuracy_score(test_y, pred))
```
