```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import pandas as pd
```

```python
import PyPDF2
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Correct relative path to the PDF based on your Jupyter folders
file_path = '../datasets after 14 sept/budget_speech.pdf'
```

```python
# Extract text from the PDF
pdf_text = ""
with open(file_path, 'rb') as file:
    reader = PyPDF2.PdfReader(file)
    for page in reader.pages:
        extracted = page.extract_text()
        if extracted:
            pdf_text += extracted + " "
```

```python
# Generate the Word Cloud
cloud = WordCloud(stopwords=STOPWORDS, max_words=100).generate(pdf_text)

plt.figure(figsize=(10, 10))
plt.imshow(cloud)
plt.axis('off') # Optional: removes the axis numbers for a cleaner look
plt.show()
```
